```python
import pandas as pd
import requests

url = "https://en.wikipedia.org/wiki/List_of_countries_and_territories_where_English_is_an_official_language"
headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(url, headers=headers)
tables = pd.read_html(response.text)

# Table 0: de jure official (includes India)
# Table 1: de facto official (includes USA, UK, Australia, NZ, Antigua & Barbuda)
df_dejure  = tables[0][['Country', 'ISO code']]
df_defacto = tables[1][['Country', 'ISO code']]

df_countries = pd.concat([df_dejure, df_defacto], ignore_index=True).drop_duplicates()
df_countries['ISO code'] = df_countries['ISO code'].str.strip()

english_speaking = set(df_countries['ISO code'])
print(len(english_speaking))
print('USA' in english_speaking, 'GBR' in english_speaking, 'IND' in english_speaking)
# -> True True True
```

```python
D1 = master_frame[
    (master_frame.country_code=='USA') &
    (master_frame.raised_amount_usd>=5000000) &
    (master_frame.raised_amount_usd<=15000000) &
    (master_frame.funding_round_type=='venture')
]
```

```python
D2 = master_frame[
    (master_frame.country_code=='GBR') &
    (master_frame.raised_amount_usd>=5000000) &
    (master_frame.raised_amount_usd<=15000000) &
    (master_frame.funding_round_type=='venture')
]

D3 = master_frame[
    (master_frame.country_code=='CHN') &
    (master_frame.raised_amount_usd>=5000000) &
    (master_frame.raised_amount_usd<=15000000) &
    (master_frame.funding_round_type=='venture')
]
```

```python
D1['sector_count'] = D1.groupby('main_sector')['main_sector'].transform('count')
D2['sector_count'] = D2.groupby('main_sector')['main_sector'].transform('count')
D3['sector_count'] = D3.groupby('main_sector')['main_sector'].transform('count')
```

```python
D1['sector_total_investment'] = D1.groupby('main_sector')['raised_amount_usd'].transform('sum')
D2['sector_total_investment'] = D2.groupby('main_sector')['raised_amount_usd'].transform('sum')
D3['sector_total_investment'] = D3.groupby('main_sector')['raised_amount_usd'].transform('sum')
```

```python
D1.groupby('main_sector').agg(
    Total_Investments=('raised_amount_usd', 'count'),
    Total_Amount=('raised_amount_usd', 'sum')
).sort_values(by='Total_Investments', ascending=False)

D2.groupby('main_sector').agg(
    Total_Investments=('raised_amount_usd', 'count'),
    Total_Amount=('raised_amount_usd', 'sum')
).sort_values(by='Total_Investments', ascending=False)

D3.groupby('main_sector').agg(
    Total_Investments=('raised_amount_usd', 'count'),
    Total_Amount=('raised_amount_usd', 'sum')
).sort_values(by='Total_Investments', ascending=False)
```

```python
table_5_1 = pd.DataFrame({
    'Country': ['USA', 'GBR', 'CHN'],
    'Top Main Sector': [
        D1['main_sector'].value_counts().sum(),
        D2['main_sector'].value_counts().sum(),
        D3['main_sector'].value_counts().sum()
    ],
    'Number of Investments': [
        D1['main_sector'].value_counts().count(),
        D2['main_sector'].value_counts().count(),
        D3['main_sector'].value_counts().count()
    ]
})

table_5_1
```
